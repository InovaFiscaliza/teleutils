```python
numeros_origem = [
    "11559699260063F0",
    "115569996200360F",
    "11557689314950F7",
    "115567981394057F",
]

numeros_destino = [    
    "4155969952",
    "1455699925",
    "4155769999",
    "1455679999",
]
```

```python
# Transforma cada item em uma tupla para representar uma linha e define o nome da coluna
df = spark.createDataFrame([(origem,destino) for origem,destino in zip(numeros_origem,numeros_destino)], ["numero_origem", "numero_destino"])

# Exibe o resultado
df.show()
```

```python
transformer = CDRTransformer(spark)
df_ = transformer._extract_phone_numbers(df)
df_.show()
```
